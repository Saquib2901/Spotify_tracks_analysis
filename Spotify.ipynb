```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
os.makedirs("figures", exist_ok=True)
print("📁 Folder 'figures' is ready!")



df = pd.read_csv('spotify_tracks.csv')

print("Shape of dataset:", df.shape)
print("Columns:", df.columns.tolist())



print("First 5 rows:")
display(df.head())

print("Dataset shape (rows, columns):")
print(df.shape)

print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nStatistical summary:")
display(df.describe())
```

```python
df = df.copy()

for c in ['key', 'mode', 'time_signature']:
    if c in df.columns:
        df[c] = df[c].replace(-1, pd.NA)
        df[c] = pd.to_numeric(df[c], errors='coerce').astype('Int64')

if 'duration_ms' in df.columns:
    df['duration_sec'] = (df['duration_ms'] / 1000).round().astype('Int64')

if 'language' in df.columns:
    df['language'] = df['language'].fillna('Unknown')

if 'track_id' in df.columns:
    df = df.drop_duplicates(subset=['track_id'])

df.shape
```

```python
# List of numerical columns to summarize
cols = ["popularity","danceability","energy","acousticness","instrumentalness",
        "speechiness","valence","loudness","tempo","duration_ms"]

# Generate individual summaries
for c in cols:
    print(f"\nStatistical Summary for: {c}\n")
    print(df[c].describe())
```

```python
plt.figure(figsize=(9,4.5))
sns.histplot(df['valence'], bins=30, color='steelblue',
             edgecolor='black', linewidth=1.1, alpha=0.65)

sns.kdeplot(df['valence'], color='darkblue', linewidth=2)

plt.title("Distribution of Valence (Musical Positivity)", fontsize=15, fontweight='bold')
plt.xlabel("Valence Score (0 = Very Sad, 1 = Very Positive)", fontsize=11)
plt.ylabel("Song Count", fontsize=11)

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,4.5))
sns.histplot(df['tempo'], bins=35, color='mediumseagreen',
             edgecolor='black', linewidth=1.1, alpha=0.65)


mean_tempo = df['tempo'].mean()
plt.axvline(mean_tempo, color='red', linestyle='--', linewidth=2.2,
            label=f"Average Tempo = {mean_tempo:.1f} BPM")

plt.title("Distribution of Track Tempos (BPM)", fontsize=15, fontweight='bold')
plt.xlabel("Tempo (Beats Per Minute)", fontsize=11)
plt.ylabel("Song Count", fontsize=11)

plt.legend(frameon=False)
plt.grid(axis='y', linestyle='--', alpha=0.32)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,4.5))

sns.violinplot(x=df['tempo'], color='mediumseagreen', linewidth=1.15)

mean_tempo = df['tempo'].mean()
median_tempo = df['tempo'].median()

plt.axvline(mean_tempo, color='red', linestyle='--', linewidth=2.2,
            label=f"Mean = {mean_tempo:.1f} BPM")
plt.axvline(median_tempo, color='navy', linestyle='-', linewidth=1.8,
            label=f"Median = {median_tempo:.1f} BPM")

plt.title("Tempo Distribution Spread (BPM Violin Plot)", fontsize=15, fontweight='bold')
plt.xlabel("Tempo (Beats Per Minute)", fontsize=11)

plt.legend(frameon=False, fontsize=10)
plt.grid(axis='x', linestyle='--', alpha=0.30)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

lang_counts = df['language'].value_counts().head(5)

plt.figure(figsize=(7,7))
colors = plt.cm.Set3(range(len(lang_counts)))   # Soft palette for visuals

plt.pie(lang_counts, labels=lang_counts.index, autopct='%1.1f%%',
        startangle=120, pctdistance=0.82,
        colors=colors, wedgeprops={'linewidth':1.2, 'edgecolor':'black'})

centre_circle = plt.Circle((0,0), 0.68, fc='white')
plt.gca().add_artist(centre_circle)

plt.title("Top 5 Languages in Dataset", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
year_counts = df['year_group'].value_counts()

plt.figure(figsize=(5,5))
plt.pie(year_counts, labels=year_counts.index, autopct='%1.1f%%', startangle=140)
centre_circle = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Song Distribution by Year Groups")
plt.tight_layout()
plt.show()
```

```python
top_langs = df['language'].value_counts().nlargest(6).index.tolist() if 'language' in df.columns else []
df_lang = df[df['language'].isin(top_langs)].copy() if top_langs else df.copy()
plt.figure(figsize=(8,6))
sns.scatterplot(data=df_lang, x='energy', y='popularity', hue='language', alpha=0.6)
plt.title('Popularity vs Energy (Top languages highlighted)')
plt.legend(loc='best', fontsize='small')
plt.tight_layout()
plt.savefig('figures/graph6_popularity_vs_energy_by_language.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='danceability', y='popularity', alpha=0.4)
sns.regplot(data=df, x='danceability', y='popularity', scatter=False, lowess=True, ci=None, color='red')
plt.title('Popularity vs Danceability')
plt.tight_layout()
plt.savefig('figures/graph5_popularity_vs_danceability.png', dpi=150)
plt.show()
```

```python
df['tempo_group'] = pd.cut(
    df['tempo'],
    bins=[0,80,110,140,200,300],
    labels=['Slow','Mid','Fast','Very Fast','Extreme']
)

plt.figure(figsize=(9,5.2))

sns.boxplot(
    x='tempo_group', y='popularity', data=df,
    palette='viridis', width=0.55,
    linewidth=1.2,  # sharper box edges
    boxprops={'alpha':0.85}
)

plt.title("Popularity Variation Across Tempo Ranges", fontsize=16, fontweight='bold')
plt.xlabel("Tempo Categories", fontsize=12)
plt.ylabel("Popularity Score", fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.28)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5.2))

sns.regplot(
    x=df['tempo'], y=df['popularity'],
    scatter_kws={'alpha':0.18, 'color':'purple', 's':14},
    line_kws={'color':'black', 'linewidth':2.7}
)

plt.title("Relationship Between Tempo and Popularity", fontsize=16, fontweight='bold')
plt.xlabel("Tempo (BPM)", fontsize=12)
plt.ylabel("Popularity Score", fontsize=12)

plt.grid(linestyle='--', alpha=0.28)
plt.tight_layout()
plt.show()
```

```python
df['duration_min'] = df['duration_ms'] / 60000

plt.figure(figsize=(9,5.2))

sns.regplot(
    x=df['duration_min'],
    y=df['popularity'],
    scatter_kws={'alpha':0.18, 'color':'brown', 's':14},
    line_kws={'color':'black', 'linewidth':2.7}
)

plt.title("Relationship Between Track Duration and Popularity", fontsize=16, fontweight='bold')
plt.xlabel("Duration (Minutes)", fontsize=12)
plt.ylabel("Popularity Score", fontsize=12)

plt.grid(linestyle='--', alpha=0.28)
plt.tight_layout()
plt.show()
```

```python
df['energy_group'] = pd.cut(
    df['energy'],
    bins=[0,0.3,0.6,1.0],
    labels=['Low Energy','Medium Energy','High Energy']
)

df['dance_group'] = pd.cut(
    df['danceability'],
    bins=[0,0.3,0.6,1.0],
    labels=['Low Dance','Medium Dance','High Dance']
)
```

```python
plt.figure(figsize=(9,5.5))

pd.crosstab(df['energy_group'], df['dance_group']).plot(
    kind='bar',
    figsize=(9,5.5),
    edgecolor='black',
    colormap='viridis',
    width=0.76
)

plt.title("Danceability vs Energy Levels (Category Comparison)", fontsize=16, fontweight='bold')
plt.xlabel("Energy Category", fontsize=12)
plt.ylabel("Track Count", fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.xticks(rotation=0, fontsize=11)  # keeps labels clean + readable
plt.tight_layout()
plt.show()
```

```python

if 'artist_name' in df.columns and 'popularity' in df.columns:
    artist_avg = df.groupby('artist_name').agg({'popularity': ['mean', 'count']})
    artist_avg.columns = ['pop_mean', 'count']
    artist_avg = artist_avg[artist_avg['count'] >= 5].sort_values('pop_mean', ascending=False).head(10)
    artist_avg = artist_avg.rename_axis('Artist').reset_index()

    plt.figure(figsize=(10,6))
    sns.barplot(x='pop_mean', y='Artist', data=artist_avg, color='skyblue')
    plt.xlabel('Average Popularity')
    plt.ylabel('Artist')
    plt.title('Top 10 Artists by Average Popularity (>=5 tracks)')
    plt.tight_layout()
    plt.savefig('figures/graph7_top_artists_avg_popularity.png', dpi=150)
    plt.show()
```

```python
year_ac = df.groupby('year')[['acousticness','instrumentalness']].mean().reset_index()

plt.figure(figsize=(9,5.2))

plt.plot(year_ac['year'], year_ac['acousticness'],
         marker='o', color='#00A6A6', linewidth=2.4, label='Acousticness')     # Teal
plt.plot(year_ac['year'], year_ac['instrumentalness'],
         marker='o', color='#D81B60', linewidth=2.4, label='Instrumentalness') # Magenta

plt.title("Acousticness & Instrumentalness Over the Years", fontsize=16, fontweight='bold')
plt.xlabel("Year", fontsize=12)
plt.ylabel("Avg Feature Value", fontsize=12)
plt.legend(frameon=False, fontsize=11)
plt.grid(linestyle='--', alpha=0.32)
plt.tight_layout()
plt.show()
```

```python
popular = df[df['popularity'] > 70]

year_popdur = popular.groupby('year')['duration_ms'].mean().reset_index()
year_popdur['duration_min'] = year_popdur['duration_ms'] / 60000

plt.figure(figsize=(8,5))
plt.plot(year_popdur['year'], year_popdur['duration_min'], marker='o', color='brown')

plt.title("Average Duration of Popular Songs Over Years", fontsize=15, weight='bold')
plt.xlabel("Year")
plt.ylabel("Average Duration (minutes)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
popular = df[df['popularity'] > 70]

year_features = popular.groupby('year')[['danceability', 'energy']].mean().reset_index()

plt.figure(figsize=(9,5.2))

plt.plot(year_features['year'], year_features['danceability'],
         marker='o', color='dodgerblue', linewidth=2.2, label='Danceability')

plt.plot(year_features['year'], year_features['energy'],
         marker='o', color='darkorange', linewidth=2.2, label='Energy')

plt.title("Danceability & Energy Trend in Popular Songs (Year-wise)", fontsize=16, fontweight='bold')
plt.xlabel("Year", fontsize=12)
plt.ylabel("Average Score", fontsize=12)
plt.legend(frameon=False, fontsize=11)
plt.grid(linestyle='--', alpha=0.32)
plt.tight_layout()
plt.show()
```
